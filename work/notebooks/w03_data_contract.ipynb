## Fields

### Features
- impressions_90d
- clicks_90d
- ctr_90d
- avg_position
- word_count

These features are known before making a content refresh decision.

### Label / Proxy
- is_declining_label (or the label provided in the dataset)

This is the value the model tries to predict.

### Context
- page_type
- month
- language

These fields provide additional information but are not the prediction target.

### Excluded
- Future traffic
- Future clicks
- Future labels

These are excluded because they would leak future information into the model and produce unrealistic performance.

## Data Limits

This dataset cannot prove that refreshing a page causes better search performance.

It only contains observed search and engagement data.

Some historical information may be incomplete, and seasonal trends cannot be fully captured using a single month's data.

The results should therefore be interpreted as decision-support rather than causal evidence.
